```python
%pip install pandas # installed pandas
import pandas as pd # imported as pd
import numpy as np # imported as np
import matplotlib.pyplot as plt # imported as plt
```

```python
# starting project 3

df = pd.read_csv('/Users/pragya.sharma/Downloads/archive-2/forest_area_km.csv')  # read the dataset 3 as a csv file
df 
```

```python
filtered_df = df[['1997', '2013']].dropna() # drop all the years except 1997 and 2017
filtered_df # display the filtered dataset
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(filtered_df['1997'], filtered_df['2013']) # create a scatter plot with (x,y) 
plt.xlabel("Forest Area in Km in 1993 per country")
plt.ylabel("Forest Area in Km in 2013 per country")
plt.title("Scatter Plot: Forest Area in Km in 1997 vs 2013")
plt.grid(True)
plt.show() # display the dataset
```

```python
# Create two subsets for clarity
increased_df = filtered_df[filtered_df['2013'] > filtered_df['1997']]
decreased_df = filtered_df[filtered_df['2013'] < filtered_df['1997']]

# Plot with color separation
plt.figure(figsize=(10, 6))
plt.scatter(increased_df['1997'], increased_df['2013'], color='green', alpha=0.7, label='Forest Increased')
plt.scatter(decreased_df['1997'], decreased_df['2013'], color='orange', alpha=0.7, label='Forest Decreased')

# Reference line (y = x)
plt.plot([filtered_df['1997'].min(), filtered_df['1997'].max()],
         [filtered_df['1997'].min(), filtered_df['1997'].max()],
         color='red', linestyle='--', label='No Change Line (y = x)')

# Labels and legend
plt.xlabel('Forest Area in 1997 (km²)')
plt.ylabel('Forest Area in 2013 (km²)')
plt.title('Forest Area Change by Country (1997 vs 2013)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
