```python
df = pd.read_csv('/Users/pragya.sharma/Downloads/co2_emission.csv')  # read the dataset as a csv file
df
```

```python
print(df.columns) # printed out the column names from the dataset
```

```python
filtered_df = df[df['Year'].isin([1997, 2013])] # filtered the dataset to only have the years 1997 and 2017
filtered_df
```

```python
filtered_df.head(500)  # displayed all the countries in 1997 and 2017
```

```python
pivot_df = df.pivot(index='Entity', columns='Year', values='Annual CO₂ emissions (tonnes )') # changed the rows to become columns
pivot_df # displayed the columns
```

```python
filtered_df = pivot_df[[1997, 2013]].dropna() # drop all the years except 1997 and 2017
filtered_df # display the filtered dataset
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(filtered_df[1997], filtered_df[2013]) # create a scatter plot with (x,y) 
plt.xlabel("CO₂ Emissions in 1997 (tonnes)")
plt.ylabel("CO₂ Emissions in 2013 (tonnes)")
plt.title("Scatter Plot: CO₂ Emissions in 1997 vs 2013")
plt.grid(True)
plt.show() # display the dataset
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
filtered_df['Cluster'] = kmeans.fit_predict(filtered_df[[1997, 2013]])

filtered_df.head(500)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(filtered_df[1997], filtered_df[2013], c=filtered_df['Cluster'], cmap='Set1', alpha=0.7)
plt.xlabel("CO₂ Emissions in 1997 (tonnes)")
plt.ylabel("CO₂ Emissions in 2013 (tonnes)")
plt.title("KMeans Clustering (3 Groups) of CO₂ Emissions")
plt.colorbar(label="Cluster (0 = Low, 2 = High)")
plt.grid(True)
plt.show()
```

```python
filtered_df = filtered_df[~filtered_df.index.isin(['World', 'Statistical differences'])]

kmeans = KMeans(n_clusters=3, random_state=42)
filtered_df['Cluster'] = kmeans.fit_predict(filtered_df[[1997, 2013]])

filtered_df.head(500)
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))
plt.scatter(filtered_df[1997], filtered_df[2013], c=filtered_df['Cluster'], cmap='Set1', alpha=0.7)
plt.xlabel("CO₂ Emissions in 1997 (tonnes)")
plt.ylabel("CO₂ Emissions in 2013 (tonnes)")
plt.title("KMeans Clustering (3 Groups) of CO₂ Emissions")
plt.colorbar(label="Cluster (0 = Low, 2 = High)")
plt.grid(True)
plt.show()

```

```python
import matplotlib.pyplot as plt

# Step 1: Filter the focused cluster (e.g., cluster 0)
focused_cluster = 0
focused_df = filtered_df[filtered_df['Cluster'] == focused_cluster].copy()

# Step 2: Apply sub-clustering on the focused cluster
sub_kmeans = KMeans(n_clusters=3, random_state=42)
focused_df['SubCluster'] = sub_kmeans.fit_predict(focused_df[[1997, 2013]])

# Step 3: Plot using SubCluster for color, and show country names
plt.figure(figsize=(12, 8))
scatter = plt.scatter(
    focused_df[1997], 
    focused_df[2013], 
    c=focused_df['SubCluster'], 
    cmap='Set1', 
    s=70, 
    alpha=0.8
)

# Add country name labels
for _, row in focused_df.iterrows():
    country_name = row.name  # assuming 'Country' is the index
    plt.text(row[1997], row[2013], country_name, fontsize=8, alpha=0.7)

plt.xlabel("CO₂ Emissions in 1997 (tonnes)")
plt.ylabel("CO₂ Emissions in 2013 (tonnes)")
plt.title(f"Zoomed-In View: Subclusters within Cluster {focused_cluster}")
plt.colorbar(scatter, label="SubCluster ID")
plt.grid(True)
plt.show()
```

```python
green_cluster = focused_df[focused_df['SubCluster'] == 0]

```

```python
plt.figure(figsize=(10, 6))
plt.scatter(green_cluster[1997], green_cluster[2013], c='green', s=60, alpha=0.8)

# Add country labels
for _, row in green_cluster.iterrows():
    plt.text(row[1997], row[2013], row.name, fontsize=8, alpha=0.7)

plt.xlabel("CO₂ Emissions in 1997 (tonnes)")
plt.ylabel("CO₂ Emissions in 2013 (tonnes)")
plt.title("Zoomed-In View: Green Cluster (SubCluster 0)")
plt.grid(True)
plt.show()
```
